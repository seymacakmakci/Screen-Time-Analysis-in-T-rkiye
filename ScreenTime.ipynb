### _Step 3: Encoding Categorical Variables_

```python
# 1. One-Hot Encoding
onehot_cols = ['occupation', 'main_device', 'peak_screenTimePeriod']
df = pd.get_dummies(df, columns=onehot_cols, dtype=int)
```

```python
print([repr(col) for col in df.columns])
```

```python
df.columns = df.columns.str.strip("'")
```

```python
print(df.columns.tolist())
```

```python
# 2. Ordinal Encoding

edu_level_map = {
    "İlkokul (Primary School)": 0,
    "Ortaokul (Middle School)": 1,
    "Lise (High School)": 2,
    "Lisans (Bachelor’s Degree)": 3,
    "Yüksek Lisans (Master’s Degree)": 4,
    "Doktora (PhD)": 5
}
df['edu_level'] = df['edu_level'].map(edu_level_map)

freq_map = {
    "Hiç": 0,
    "Nadiren": 1,
    "Bazen": 2,
    "Sık": 3,
    "Çok sık": 4
}
freq_cols = [
    'ScreenUse [screenUse_WorkStudy]',
    'ScreenUse [screenUse_SocialMedia]',
    'ScreenUse [screenUse_Streaming]',
    'ScreenUse [screenUse_Gaming]',
    'ScreenUse [screenUse_OnlineShopping]',
    'ScreenUse [screenUse_ReadingNews]'
]

freq_map = {
    "Hiç": 0,
    "Az": 1,
    "Orta": 2,
    "Çok": 3,
    "Çok Fazla": 4
}

for col in freq_cols:
    df[col] = df[col].map(freq_map)

emotion_map = {
    "Olumsuz": 0,
    "Nötr": 1,
    "Olumlu": 2
}

df['EmotionalState_AfterUse'] = df['EmotionalState_AfterUse'].map(emotion_map)
```

```python
# 3. Label Encoding (remaining categorical columns)
from sklearn.preprocessing import LabelEncoder


label_encoders = {}

exclude_cols = onehot_cols + freq_cols + ['edu_level', 'emotionalState_AfterUse']

categorical_cols = df.select_dtypes(include='object').columns.tolist()

label_cols = [col for col in categorical_cols if col not in exclude_cols]

for col in label_cols:
    df[col] = df[col].astype(str)
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le
    print(f"{col} -> {list(le.classes_)}")
```
